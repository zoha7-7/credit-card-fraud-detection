```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from imblearn.over_sampling import SMOTE
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler

print("All libraries imported successfully!")
```

```python
from pathlib import Path

try:
    import matplotlib.pyplot as plt
    import numpy as np
    import pandas as pd
    import seaborn as sns
    from imblearn.over_sampling import SMOTE
    from sklearn.impute import SimpleImputer
    from sklearn.model_selection import train_test_split
    from sklearn.preprocessing import RobustScaler
except ImportError as exc:
    raise ImportError(
        "Member 2 could not import a required Python package. "
        "If the original error says Windows Application Control blocked a DLL, "
        "this is a Python environment/security-policy issue, not a dataset error. "
        "Use an approved Python environment or ask your administrator to allow "
        f"the required library. Original error: {exc}"
    ) from exc


BASE_DIR = Path("/content").resolve()
DATA_PATH = BASE_DIR / "creditcard.csv"
REPORT_DIR = BASE_DIR / "reports"
GRAPH_DIR = BASE_DIR / "graphs"
RANDOM_STATE = 42
TEST_SIZE = 0.2
TARGET = "Class"

REPORT_DIR.mkdir(exist_ok=True)
GRAPH_DIR.mkdir(exist_ok=True)
sns.set_theme(style="whitegrid")

if not DATA_PATH.is_file():
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}\n"
        "Place creditcard.csv in the same folder as member2.py."
    )

# Load and validate the transaction data.
df = pd.read_csv(DATA_PATH)
if TARGET not in df.columns:
    raise ValueError(f"Dataset must contain the target column {TARGET!r}.")

if not df[TARGET].dropna().isin([0, 1]).all():
    raise ValueError(f"{TARGET} must contain only 0 (legitimate) and 1 (fraud).")

feature_columns = [column for column in df.columns if column != TARGET]
non_numeric_columns = df[feature_columns].select_dtypes(exclude=np.number).columns.tolist()
if non_numeric_columns:
    raise ValueError(
        "All feature columns must be numeric for this workflow. "
        f"Non-numeric columns: {non_numeric_columns}"
    )

original_rows = len(df)
missing_report = (
    df.isna()
    .sum()
    .rename("missing_values")
    .to_frame()
    .assign(missing_percent=lambda result: result["missing_values"] / len(df) * 100)
)
missing_report.to_csv(REPORT_DIR / "member2_missing_values.csv")

# Rows without a target cannot be used for supervised training. Remove these,
# then remove exact duplicate records before splitting to prevent leakage.
missing_target_rows = int(df[TARGET].isna().sum())
df = df.dropna(subset=[TARGET]).copy()
duplicate_rows = int(df.duplicated().sum())
df = df.drop_duplicates().copy()
df[TARGET] = df[TARGET].astype("int8")

if df[TARGET].nunique() != 2:
    raise ValueError("Both classes (0 and 1) must remain after cleaning.")

X = df[feature_columns]
y = df[TARGET]

# Detect extreme values using the IQR rule for reporting. Do not delete these
# rows: extreme transactions may be genuine fraud signals. RobustScaler below
# scales features using medians and interquartile ranges.
q1 = X.quantile(0.25)
q3 = X.quantile(0.75)
iqr = q3 - q1
outlier_mask = X.lt(q1 - 1.5 * iqr) | X.gt(q3 + 1.5 * iqr)
outlier_report = pd.DataFrame(
    {
        "outlier_values": outlier_mask.sum(),
        "outlier_percent": outlier_mask.mean() * 100,
        "handling": "Preserved; RobustScaler limits sensitivity to extreme values",
    }
)
outlier_report.to_csv(REPORT_DIR / "member2_outlier_report.csv")

# Split before learning imputation or scaling parameters. The test set remains
# untouched by SMOTE and is suitable for an honest final model evaluation.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y,
)

# Fit preprocessing on training features only to avoid data leakage.
imputer = SimpleImputer(strategy="median", keep_empty_features=True)
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

scaler = RobustScaler()
X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)

# Balance training data only. Keep the original class distribution in the test set.
smote = SMOTE(random_state=RANDOM_STATE)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train_scaled, y_train)

train_output = pd.DataFrame(X_train_balanced, columns=feature_columns)
train_output[TARGET] = np.asarray(y_train_balanced, dtype="int8")
test_output = pd.DataFrame(X_test_scaled, columns=feature_columns)
test_output[TARGET] = y_test.to_numpy(dtype="int8")

train_path = BASE_DIR / "member2_train_balanced.csv"
test_path = BASE_DIR / "member2_test.csv"
train_output.to_csv(train_path, index=False)
test_output.to_csv(test_path, index=False)

# Save class counts before and after resampling.
class_distribution = pd.DataFrame(
    {
        "stage": ["cleaned dataset", "training before SMOTE", "training after SMOTE", "test set"],
        "legitimate (0)": [
            int(y.value_counts().get(0, 0)),
            int(y_train.value_counts().get(0, 0)),
            int(pd.Series(y_train_balanced).value_counts().get(0, 0)),
            int(y_test.value_counts().get(0, 0)),
        ],
        "fraud (1)": [
            int(y.value_counts().get(1, 0)),
            int(y_train.value_counts().get(1, 0)),
            int(pd.Series(y_train_balanced).value_counts().get(1, 0)),
            int(y_test.value_counts().get(1, 0)),
        ],
    }
)
class_distribution["fraud_percent"] = (
    class_distribution["fraud (1)"]
    / (class_distribution["legitimate (0)"] + class_distribution["fraud (1)"])
    * 100
)
class_distribution["legitimate_to_fraud_ratio"] = (
    class_distribution["legitimate (0)"] / class_distribution["fraud (1)"]
)
class_distribution.to_csv(REPORT_DIR / "member2_class_distribution.csv", index=False)

overview = pd.DataFrame(
    {
        "original_rows": [original_rows],
        "rows_with_missing_target_removed": [missing_target_rows],
        "duplicate_rows_removed": [duplicate_rows],
        "cleaned_rows_before_split": [len(df)],
        "training_rows_after_smote": [len(train_output)],
        "test_rows": [len(test_output)],
        "feature_count": [len(feature_columns)],
        "missing_feature_values_after_imputation": [
            int(train_output[feature_columns].isna().sum().sum()
                + test_output[feature_columns].isna().sum().sum())
        ],
    }
)
overview.to_csv(REPORT_DIR / "member2_preparation_overview.csv", index=False)

# Visualize the original class imbalance and the training classes after SMOTE.
fig, axes = plt.subplots(1, 2, figsize=(11, 5))
before_counts = y.value_counts().reindex([0, 1], fill_value=0)
after_counts = pd.Series(y_train_balanced).value_counts().reindex([0, 1], fill_value=0)
sns.barplot(x=["Legitimate", "Fraud"], y=before_counts.values, ax=axes[0])
axes[0].set(title="Class distribution before SMOTE", ylabel="Transactions", xlabel="")
sns.barplot(x=["Legitimate", "Fraud"], y=after_counts.values, ax=axes[1])
axes[1].set(title="Training class distribution after SMOTE", ylabel="Transactions", xlabel="")
fig.tight_layout()
fig.savefig(GRAPH_DIR / "member2_class_balance.png", dpi=150)
plt.close(fig)

# Show the Amount distribution by class when that common dataset column exists.
if "Amount" in df.columns:
    amount_data = df[["Amount", TARGET]].copy()
    amount_data["log1p(Amount)"] = np.log1p(amount_data["Amount"].clip(lower=0))
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.boxplot(data=amount_data, x=TARGET, y="log1p(Amount)", ax=ax)
    ax.set(
        title="Transaction amount by class (log1p scale)",
        xlabel="Class (0=legitimate, 1=fraud)",
        ylabel="log(1 + Amount)",
    )
    fig.tight_layout()
    fig.savefig(GRAPH_DIR / "member2_amount_by_class.png", dpi=150)
    plt.close(fig)

print("Member 2 preprocessing completed.")
print(f"Rows removed as duplicates: {duplicate_rows:,}")
print(f"Rows removed because Class was missing: {missing_target_rows:,}")
print(f"Training rows after SMOTE: {len(train_output):,}")
print(f"Test rows (not resampled): {len(test_output):,}")
print(f"Training data: {train_path}")
print(f"Test data: {test_path}")
print(f"Reports: {REPORT_DIR}")
print(f"Graphs: {GRAPH_DIR}")
```

### Step 4: Model Training
We will train a Random Forest Classifier on our balanced training dataset (`member2_train_balanced.csv`). To keep execution times reasonable on Colab, we'll configure a solid baseline with `n_estimators=100` and utilize all CPU cores (`n_jobs=-1`).

```python
from sklearn.ensemble import RandomForestClassifier
import joblib

# 1. Load balanced training data
print("Loading balanced training data...")
train_df = pd.read_csv(train_path)
X_train_b = train_df.drop(columns=[TARGET])
y_train_b = train_df[TARGET]

# 2. Initialize and train the Random Forest Classifier
print("Training Random Forest Classifier on balanced data (this may take a few minutes)...")
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    verbose=1
)
rf_model.fit(X_train_b, y_train_b)

# 3. Save the trained model
model_dir = BASE_DIR / "models"
model_dir.mkdir(exist_ok=True)
model_path = model_dir / "member2_random_forest.joblib"
joblib.dump(rf_model, model_path)

print(f"Model training completed successfully and saved to: {model_path}")
```
